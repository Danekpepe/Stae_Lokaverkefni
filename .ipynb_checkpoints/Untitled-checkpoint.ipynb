```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.sparse import diags, csr_matrix
from Gemini_part1 import heatwave2, solve_ua, PoissonEq3, plot_comparison_table, theoretical_energy, compute_energy
```

```python
L = 1
T = 1/20
N = 6
M = 4

def f(x):
    return np.ones_like(x) * 100

def psi(t):
    return np.zeros_like(t)

def phi(t):
    return np.zeros_like(t)
```

```python
print(heatwave2(L, T, N, M, psi, phi, f))
```

```python
def f_0(x):
    return 1 - np.abs(x- 1)

L_0 = 2
T_0 = 1
N_0 = 10
M_0 = 100

hw = heatwave2(L_0,T_0,N_0,M_0,psi,phi,f_0)
```

```python
print("HW:", hw[-1,:])
```

```python
sol = []
for j in range(M_0+1):
    for i in range(N_0+1):    
        sol.append(solve_ua(L_0/(N_0)*i,T_0/(M_0)*j ,2))

ua = np.array(sol).reshape(M_0+1,N_0+1)
```

```python
print("UA:", ua[-1,:])
```

```python
x_vals = [f"x = {v}" for v in np.linspace(0, L_0, N_0+1)[-5:]]
t_vals = [f"t = {v}" for v in np.linspace(0, T_0, M_0+1)[-5:]]


plot_comparison_table(hw[-5:,-5:], ua[-5:,-5:], x_vals, t_vals, title="Samanburður Gilda")
```

```python
t = np.linspace(0,1,101)

E_comp = compute_energy(hw, 1/100)
print(E_comp[:5])

plt.plot(t,E_comp)
```

```python
E_theor = []
for i in range(101):
    E_theor.append(theoretical_energy(1/101 * i,2))

print(E_theor[:5])

plt.plot(t,E_theor)
```
